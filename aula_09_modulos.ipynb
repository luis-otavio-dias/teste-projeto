# Módulos

Os módulos são geralmente usados para expandir as capacidades da linguagem, carregar funcionalidades e permitir que seu programa possa ser segmentado em unidades menores. Eles são simplesmente arquivos com a extensão ".py".

Para usar essas funções em um código, você deve importar o módulo usando o comando `import`. No exemplo abaixo, usamos o módulo `math` para operações matemáticas.

```python
import math
```

```python
print(2 ** 2)

# 2 elevado a 3
print(math.pow(2, 3))

# arredondando valores
print(math.ceil(2.5))
print(math.trunc(2.5))
```

### Excluindo arquivos e pastas

Para excluir um arquivo, você deve importar o módulo `os` e executar sua função `os.remove()`. É sempre uma boa prática, antes de excluir um arquivo, verificar se ele existe usando `os.path.exists()`.

```python
# Verifica se o arquivo existe para depois remover
import os

if os.path.exists("arquivo.txt"):
  os.remove("arquivo.txt")
else:
  print("O arquivo não existe")
```

Para excluir uma pasta, usamos a função `os.rmdir()` do módulo `os`. Note que a pasta deve estar vazia e o caminho deve estar correto.

```python
import os

os.rmdir("folderExemplo")
```

### Explorando módulos embutidos

Duas funções muito úteis para explorar módulos são `dir()` e `help()`. Com `dir()`, você pode procurar quais funções são implementadas. Com `help()`, você lê a descrição de como usar uma função específica.

```python
import math

# exibe todas as funções de um módulo
print(dir(math))
```

```python
help(math)
```

### Escrevendo seus próprios módulos

Para criar um módulo próprio, basta criar um arquivo `.py` (como o `calculadora.py`) e importá-lo pelo nome do arquivo. Abaixo, importamos o módulo personalizado que contém as quatro operações básicas.

```python
import calculadora

print(calculadora.soma(10, 30))

print(calculadora.subtracao(30, 10))
print(calculadora.multiplicacao(10, 30))
print(calculadora.divisao(30, 10))
```

# Pacotes

Os pacotes são diretórios que contêm vários módulos. Para que o Python reconheça uma pasta como pacote, ela deve conter um arquivo especial chamado `__init__.py`.

Podemos importar módulos de pacotes de duas formas: usando apenas `import` ou usando `from`.

```python
import matematica.calculadora

matematica.calculadora.soma(2, 3)

matematica.calculadora.subtracao(3, 1)
```

```python
from matematica import calculadora,geometria

print(calculadora.soma(2, 3))

print(geometria.areaQuadrado(3))
```

```python
from calculadora import divisao, multiplicacao

print(divisao(10, 2))
print(multiplicacao(5, 3))
```

### Apelidos (Alias)

Para evitar conflitos de nomes ou simplificar o código, podemos dar apelidos aos módulos ou funções usando a palavra-chave `as`.

```python
from matematica import geometria as g
from matematica import calculadora as calc

print(g.areaQuadrado(3))

print(calc.soma(2, 2))
```

### Desafio: Pares Entre Cinco Números

Programa para ler 5 valores e contar quantos são pares.

```python
pares = 0

# Loop para ler 5 valores
for i in range(5):
    valor = int(input())
    if valor % 2 == 0:
        pares += 1

print(f"{pares} valores pares")
```

### Desafio: Seis Números Ímpares

Leia um valor inteiro X e apresente os 6 valores ímpares consecutivos a partir de X.

```python
x = int(input())

# Se X for par, o primeiro ímpar é X + 1
# Se X for ímpar, o primeiro ímpar é o próprio X
if x % 2 == 0:
    inicio = x + 1
else:
    inicio = x

# Loop para imprimir os 6 ímpares
for i in range(6):
    print(inicio + (i * 2))
```
